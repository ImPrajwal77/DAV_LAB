```python
grade_info=pd.DataFrame({'grade':['A','B','C','D','F',],
                         'grade_description':[
                         'excellent',
                         'Good',
                         'Average',
                         'Below Average',
                         'Fail'
                        ]
                     })
print(grade_info)
```
